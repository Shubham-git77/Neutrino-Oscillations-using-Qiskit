```python
import numpy as np
from qiskit import QuantumCircuit, QuantumRegister, ClassicalRegister, transpiler
from qiskit.primitives import Sampler
from qiskit_aer import Aer
import matplotlib.pyplot as plt
from qiskit.circuit.library import HamiltonianGate



m = 1
E = 1
h = 1
qr = QuantumRegister(2)
cr = ClassicalRegister(2)
probabilities = []
time = np.linspace(0, 10, 100)

for t in time:
    circuit = QuantumCircuit(qr, cr)
    circuit.x(qr[0])
    circuit.u(np.pi / 2, 0, 0, qr[0])
    phi = (m**2 * t) / (2 * E * h)



    circuit.p(phi, qr[0])
    circuit.u(-np.pi / 2, 0, 0, qr[0])
    circuit.measure(qr, cr)

    simulator = Aer.get_backend('qasm_simulator')
    compiled_circuit = transpile(circuit, simulator)

    sampler = Sampler()
    job = sampler.run(compiled_circuit)
    result = job.result()
    quasi_dists = result.quasi_dists
    counts = quasi_dists[0].binary_probabilities()


    probability_01 = counts.get('01', 0)
    probabilities.append(probability_01)


plt.plot(time, probabilities, marker='o')
plt.xlabel("Time")
plt.ylabel("Probability of'01'")
plt.show()
```

```python
# prompt: python code for neutrino oscillations probability

import numpy as np
import matplotlib.pyplot as plt

def neutrino_oscillation_probability(m, E, h, time_steps):
    """
    Simulates neutrino oscillations and plots the probability of the initial state over time.

    Args:
        m: Mass difference squared (in eV^2)
        E: Neutrino energy (in eV)
        h: Reduced Planck constant (in eV s)
        time_steps: Array of time values (in seconds)
    """
    probabilities = []
    for t in time_steps:
        phi = (m * t) / (2 * E * h)  # Phase factor
        probability_00 = np.cos(phi)**2 # Probability of the initial state
        probabilities.append(probability_00)

    plt.plot(time_steps, probabilities)
    plt.xscale('log')
    plt.xlabel("Time(log scale) (s)")
    plt.ylabel("Probability of initial state")
    plt.title("Neutrino Oscillation Probability")
    plt.show()

# Example usage:
m = 100  # Mass difference squared (example value)
E = 1e6   # Neutrino energy (example value)
h = 4.135667696e-15 # Reduced Planck constant in eV*s
time_steps = np.linspace(0, 10, 100) # Time steps for the simulation

neutrino_oscillation_probability(m, E, h, time_steps)
```

```python
from qiskit import QuantumCircuit
from qiskit_aer import Aer
from qiskit.execute_function import execute
import numpy as np
import matplotlib.pyplot as plt

# Function to create Bell state |Φ+>
def create_bell_state():
    qc = QuantumCircuit(2)
    qc.h(0)
    qc.cx(0, 1)
    return qc

# Add measurement in rotated basis
def add_measurement(qc, theta_a, theta_b):
    circ = qc.copy()
    circ.ry(-2*theta_a, 0)   # Alice
    circ.ry(-2*theta_b, 1)   # Bob
    circ.measure_all()
    return circ

# Run experiment and return counts + expectation value
def correlation(theta_a, theta_b, shots=4096):
    qc = create_bell_state()
    circ = add_measurement(qc, theta_a, theta_b)
    backend = Aer.get_backend('qasm_simulator')
    result = execute(circ, backend, shots=shots).result()
    counts = result.get_counts()

    # Expectation value E = P(00)+P(11) - P(01) - P(10)
    exp_val = 0
    for outcome, count in counts.items():
        if outcome in ['00', '11']:
            exp_val += count
        else:
            exp_val -= count
    return counts, exp_val / shots

# Measurement angles
angles = {
    "A": 0,
    "A'": np.pi/4,
    "B": np.pi/8,
    "B'": -np.pi/8
}

# Run experiments
settings = [("A","B"), ("A","B'"), ("A'","B"), ("A'","B'")]
results = {}

for s in settings:
    counts, exp_val = correlation(angles[s[0]], angles[s[1]])
    results[s] = (counts, exp_val)

# Plot bar charts
fig, axs = plt.subplots(2, 2, figsize=(10, 8))
axs = axs.ravel()

for i, s in enumerate(settings):
    counts, exp_val = results[s]
    axs[i].bar(counts.keys(), counts.values(), color="skyblue", edgecolor="black")
    axs[i].set_title(f"{s[0]} vs {s[1]} (E = {exp_val:.2f})")
    axs[i].set_xlabel("Outcome")
    axs[i].set_ylabel("Counts")

plt.tight_layout()
plt.show()

# Compute CHSH parameter
E_AB   = results[("A","B")][1]
E_ABp  = results[("A","B'")][1]
E_ApB  = results[("A'","B")][1]
E_ApBp = results[("A'","B'")][1]
S = E_AB + E_ABp + E_ApB - E_ApBp

print("CHSH S =", S)
```
